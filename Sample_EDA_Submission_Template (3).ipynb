## 3. ***Data Wrangling***

#handling missing values(filling with mean for numerical , mode for categorical values)


### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df.fillna(df.mean(numeric_only=True),inplace=True)
df.fillna(df.mode().iloc[0],inplace=True)
df.isnull().sum()
````

````python
df['member_birth_year']=df['member_birth_year'].astype(int)
df=df.sort_values(by='member_birth_year')
print(df['member_birth_year'])
````

````python
df.dtypes
````

````python
print("unique values for columns(bike_id,user_type,start_station_name,end_station_name)")
for x in ['bike_id','user_type', 'start_station_name', 'end_station_name']:
    values=df[x].unique()
    print(f'{x}: {values}')
````
